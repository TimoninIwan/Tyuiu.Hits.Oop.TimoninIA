[Создать базовый класс Notification в C#, который будет представлять уведомления пользователям. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Notification: • Атрибуты: ID уведомления (NotificationId), Текст уведомления (MessageText), Тип уведомления (Type). • Методы: o DisplayNotification(): метод для отображения уведомления пользователю. o SendNotification(): метод для отправки уведомления. o GetNotificationDetails(): метод для получения деталей уведомления.

Требования к производным классам:

EmailУведомление (EmailNotification): Должно содержать дополнительные атрибуты, такие как Адрес электронной почты (EmailAddress). Метод SendNotification() должен быть переопределен для отправки уведомления по электронной почте.
SMSУведомление (SMSNotification): Должно содержать дополнительные атрибуты, такие как Номер телефона (PhoneNumber). Метод SendNotification() должен быть переопределен для отправки уведомления через SMS.
PushУведомление (PushNotification) (если требуется третий класс): Должно содержать дополнительные атрибуты, такие как Платформа (Platform, например, iOS или Android). Метод DisplayNotification() должен быть переопределен для отображения уведомления на мобильной платформе.]

```
public abstract class Notification
{

    public int NotificationId { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }

    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine(MessageText);
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Уведомление {NotificationId} отправлено.");
    }

    public virtual void GetNotificationDetails()
    {
        Console.WriteLine("Детали уведомления:");
        Console.WriteLine($"ID: {NotificationId} | Тип: {Type} | Сообщение: {MessageText}");
    }

    public static void SendAndDisplayNotificationGroup(Notification[] notifications)
    {
        Console.WriteLine("Отправка связанных уведомлений:");
        foreach (var notification in notifications)
        {
            notification.SendNotification();
            notification.DisplayNotification();
            Console.WriteLine();
        }
    }
}

public interface IDelayedNotification
{
    int DelayTime {get; set; }

    public void DelayOn(int minutes);

    public void DeleteDelay();
}

public class EmailNotification : Notification
{
    private string emailAddress;
    public string EmailAddress
    {
        get { return emailAddress; }
        set { emailAddress = value; }
    }

    
    public EmailNotification(string emailAdress, int notificationId, string messageText, string type) 
        : base(notificationId, messageText, type)
    {
        EmailAddress = emailAdress;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"На почте {emailAddress} новое уведомление.");
    }
}

public class CorporateEmailNotification : EmailNotification
{
    private string department;

    public string Department
    {
        get { return department; }
        set { department = value; }
    }

    public CorporateEmailNotification(string emailAddress, int notificationId, string messageText, string type, string department)
        : base(emailAddress, notificationId, messageText, type)
    {
        Department = department;
    }

    public void CorpNotification()
    {
        Console.WriteLine($"Корпоративное уведомление для отдела {Department}.");
    }

    public void CorpNotification(string sender)
    {
        Console.WriteLine($"Корпоративное уведомление для отдела {Department}, отправитель: {sender}.");
    }

    public void CorpNotification(int minutes)
    {
        Console.WriteLine($"Корпоративное уведомление для отдела {Department}, необходимо отчитаться о получении уведомления в течение {minutes} минут.");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        CorpNotification();
    }
}

public class SMSNotification : Notification
{
    private string phoneNumber;
    public string PhoneNumber
    {
        get { return phoneNumber; }
        set { phoneNumber = value; }
    }

    public SMSNotification(string phoneNumber, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"SMS отправлено на номер {phoneNumber}.");
    }
}

public class PushNotification : Notification, IDelayedNotification
{
    private string platform;
    public string Platform
    {
        get { return platform; }
        set { platform = value; }
    }

    public int DelayTime {get; set; }

    public PushNotification(string platform, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        Platform = platform;
    }

    public void DelayOn(int minutes)
    {
        DelayTime += minutes;
    }

    public void DeleteDelay()
    {
        DelayTime = 0;
    }

    public override void SendNotification()
    {
        if (DelayTime > 0)
        {
            Console.WriteLine($"Установлена задержка уведомления на {DelayTime} минут.");
            Console.WriteLine("Время прошло.");
        }
        Console.WriteLine($"Push уведомление отправлено на платформу {Platform}.");
    }
}

public class NotificationCollection<T> where T : Notification
{
    private List<T> _notifications = new List<T>();

    public void Add(T notification)
    {
        _notifications.Add(notification);
    }

    public void Remove(T notification)
    {
        _notifications.Remove(notification);
    }

    public void DisplayNotifications()
    {
        foreach (var notification in _notifications)
        {
            notification.SendNotification();
            notification.DisplayNotification();
            Console.WriteLine();
        }
    }
}

EmailNotification emailNotification = new EmailNotification(
    "timalvik2017@mail.ru", 
    1, 
    "Ваш заказ прибыл в ПВЗ.", 
    "Входящие"
);

SMSNotification smsNotification = new SMSNotification(
    "+79504895766", 
    2, 
    "Не забудьте оплатить товар. Сумма: 2000р.", 
    "Оповещение"
);

PushNotification pushNotification = new PushNotification(
    "iOS", 
    3, 
    "Ваш заказ готов к выдаче.", 
    "Оповещение"
);
pushNotification.DelayOn(30);

NotificationCollection<Notification> notifications = new NotificationCollection<Notification>();
notifications.Add(emailNotification);
notifications.Add(pushNotification);
notifications.Add(smsNotification);

notifications.DisplayNotifications();

Console.WriteLine();

CorporateEmailNotification corpotateEmail = new CorporateEmailNotification("hrNumber2@company.com", 4, 
    "Собрание в 16:00.", 
    "Meeting", 
    "HR"
);

CorporateEmailNotification corpotateEmail2 = new CorporateEmailNotification("hr413@company.com", 5, 
    "Получены обновленные данные по планированию бюджета.", 
    "Update", 
    "финансов"
);

corpotateEmail.CorpNotification("Начальник отдела Тимонин И.A");

corpotateEmail2.CorpNotification(45);
```
