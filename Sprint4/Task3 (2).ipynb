```
public abstract class Notification
{

    public int NotificationId { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }
    public DateTime CreatedDate { get; set; }
    public int Priority { get; set; }
    public bool IsRead { get; set; }
    public string Sender { get; set; }

    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
        CreatedDate = DateTime.Now;
        Priority = 1; 
        IsRead = false; 
        Sender = "System";
    }

    public virtual void MarkAsRead()
    {
        IsRead = true;
        Console.WriteLine($"Уведомление {NotificationId} помечено как прочитанное.");
    }

    public virtual void SetPriority(int priority)
    {
        Priority = Math.Clamp(priority, 1, 5);
        Console.WriteLine($"Приоритет уведомления {NotificationId} установлен: {Priority}.");
    }

    public virtual void UpdateMessage(string newMessage)
    {
        MessageText = newMessage;
        Console.WriteLine($"Сообщение уведомления {NotificationId} обновлено.");
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine(MessageText);
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Уведомление {NotificationId} отправлено.");
    }

    public virtual void GetNotificationDetails()
    {
        Console.WriteLine("Детали уведомления:");
        Console.WriteLine($"ID: {NotificationId} | Тип: {Type} | Отправитель: {Sender}");
        Console.WriteLine($"Приоритет: {Priority} | Прочитано: {IsRead} | Создано: {CreatedDate}");
        Console.WriteLine($"Сообщение: {MessageText}");
    }

}

public interface IDelayedNotification
{
    int DelayTime {get; set; }

    public void DelayOn(int minutes);

    public void DeleteDelay();
}

public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }
    public string Subject { get; set; }
    public string AttachmentPath { get; set; }

    
    public EmailNotification(string emailAddress, int notificationId, string messageText, string type, string subject = "Без темы") 
        : base(notificationId, messageText, type)
    {
        EmailAddress = emailAddress;
        Subject = subject; 
        AttachmentPath = string.Empty;
    }

    public void AddAttachment(string filePath)
    {
        AttachmentPath = filePath;
        Console.WriteLine($"Вложение к письму добавлено: {filePath}.");
    }
    public override void SendNotification()
    {
        Console.WriteLine($"Email отправлен на адрес {EmailAddress} с темой '{Subject}'.");
        if (!string.IsNullOrEmpty(AttachmentPath))
            {
                Console.WriteLine($"Прикрепленный файл: {AttachmentPath}.");
            }
    }
}

public class CorporateEmailNotification : EmailNotification
{
    public string Department { get; set; }
    public bool RequiresConfirmation { get; set; }


    public CorporateEmailNotification(string emailAddress, int notificationId, 
        string messageText, string type, string department)
        : base(emailAddress, notificationId, messageText, type)
    {
        Department = department;
        RequiresConfirmation = true;
    }


    public void SendUrgent()
    {
        Priority = 5;
        Console.WriteLine($"СРОЧНОЕ уведомление для отдела {Department}.");
    }

    public void RequestConfirmation()
    {
        if (RequiresConfirmation)
        {
            Console.WriteLine($"Запрошено подтверждение получения для отдела {Department}.");
        }
    }

    public void CorpNotification()
    {
        Console.WriteLine($"Корпоративное уведомление для отдела {Department}.");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        CorpNotification();
        if (RequiresConfirmation) RequestConfirmation();
    }
}

public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }

    public SMSNotification(string phoneNumber, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"SMS отправлено на номер {PhoneNumber}.");
    }
}

public class PushNotification : Notification, IDelayedNotification
{
    public string Platform { get; set; }
    public int DelayTime {get; set; }
    public string Sound { get; set; }

    public PushNotification(string platform, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        Platform = platform;
        Sound = "default";
    }

    public void SetSound(string soundName)
    {
        Sound = soundName;
        Console.WriteLine($"Звук пуш-уведомления установлен: {soundName}.");
    }

    void IDelayedNotification.DelayOn(int minutes)
    {
        DelayTime += minutes;
    }

     void IDelayedNotification.DeleteDelay()
    {
        DelayTime = 0;
    }

    public override void SendNotification()
    {
        if (DelayTime > 0)
        {
            Console.WriteLine($"Установлена задержка уведомления на {DelayTime} минут.");
            Console.WriteLine("Время прошло.");
        }
        Console.WriteLine($"Push уведомление отправлено на платформу {Platform}.");
    }
}

public class NotificationCollection<T> where T : Notification
{
    private List<T> _notifications = new List<T>();

    public void Add(T notification)
    {
        _notifications.Add(notification);
    }

    public void Remove(T notification)
    {
        _notifications.Remove(notification);
    }

     public IEnumerable<T> GetAll()
    {
        return _notifications;
    }

    public void DisplayNotifications()
    {
        foreach (var notification in _notifications)
        {
            notification.SendNotification();
            notification.DisplayNotification();
            Console.WriteLine();
        }
    }
}

public interface INotificationService
{
    void Send(Notification notification);
    void LogNotification(Notification notification);
}

public class NotificationService : INotificationService
{
    public void Send(Notification notification)
    {
        notification.SendNotification();
        LogNotification(notification);
    }

    public void LogNotification(Notification notification)
    {
        Console.WriteLine($"[Log] {DateTime.Now}: Уведомление {notification.NotificationId} отправлено через {notification.Type}.");
    }
}

public class NotificationManager
{
    private readonly INotificationService notificationService;
    
    public NotificationCollection<Notification> Notifications { get; set; }

    public NotificationManager(INotificationService notificationService)
    {
        this.notificationService = notificationService;
        Notifications = new NotificationCollection<Notification>();
    }

    public void SendAllNotifications()
    {
        foreach (var notification in Notifications.GetAll())
        {
            this.notificationService.Send(notification);
            Console.WriteLine();
        }
    }

    public void AddNotification(Notification notification)
    {
        Notifications.Add(notification);
        Console.WriteLine($"Уведомление {notification.NotificationId} добавлено в менеджер.");
    }
}

        var notificationService = new NotificationService();        
        var manager = new NotificationManager(notificationService);


        var push = new PushNotification("iOS", 2, "Push сообщение", "Push");
        push.SetSound("Clock"); 
        var delayedPush = (IDelayedNotification)push;
        delayedPush.DelayOn(30);
        manager.AddNotification(push);

        var email = new EmailNotification("timalvik2017@mail.com", 1, "Тестовое сообщение", "Email")
        {
            Subject = "Важное уведомление"
        };
        email.AddAttachment(@"C:\documents\file.pdf");
        manager.AddNotification(email);
        Console.WriteLine();


        Console.WriteLine();
        Console.WriteLine("=== Отправка всех уведомлений ===");
        manager.SendAllNotifications(); 
```
