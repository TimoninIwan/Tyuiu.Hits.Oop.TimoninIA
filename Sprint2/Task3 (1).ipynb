```
abstract class Notification
{

    public int NotificationId { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }

    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine(MessageText);
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Уведомление {NotificationId} отправлено.");
    }

    public virtual void GetNotificationDetails()
    {
        Console.WriteLine("Детали уведомления:");
        Console.WriteLine($"ID: {NotificationId} | Тип: {Type} | Сообщение: {MessageText}");
    }

    public static void SendAndDisplayNotificationGroup(Notification[] notifications)
    {
        Console.WriteLine("Отправка связанных уведомлений:");
        foreach (var notification in notifications)
        {
            notification.SendNotification();
            notification.DisplayNotification();
            Console.WriteLine();
        }
    }
}

interface IDelayedNotification
{
    int DelayTime {get; set; }

    public void DelayOn(int minutes);

    public void DeleteDelay();
}

class EmailNotification : Notification
{
    private string emailAddress;
    public string EmailAddress
    {
        get { return emailAddress; }
        set { emailAddress = value; }
    }

    
    public EmailNotification(string emailAdress, int notificationId, string messageText, string type) 
        : base(notificationId, messageText, type)
    {
        EmailAddress = emailAdress;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"На почте {emailAddress} новое уведомление.");
    }
}

class CorporateEmailNotification : EmailNotification
{
    private string department;

    public string Department
    {
        get { return department; }
        set { department = value; }
    }

    public CorporateEmailNotification(string emailAddress, int notificationId, string messageText, string type, string department)
        : base(emailAddress, notificationId, messageText, type)
    {
        Department = department;
    }

    public void CorpNotification()
    {
        Console.WriteLine($"Уведомление для сотрудника {Department}.");
    }

    public override void SendNotification()
    {
        base.SendNotification();
        CorpNotification();
    }
}

class SMSNotification : Notification
{
    private string phoneNumber;
    public string PhoneNumber
    {
        get { return phoneNumber; }
        set { phoneNumber = value; }
    }

    public SMSNotification(string phoneNumber, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"SMS отправлено на номер {phoneNumber}.");
    }
}

class PushNotification : Notification, IDelayedNotification
{
    private string platform;
    public string Platform
    {
        get { return platform; }
        set { platform = value; }
    }

    public int DelayTime {get; set; }

    public PushNotification(string platform, int notificationId, string messageText, string type)
         : base(notificationId, messageText, type)
    {
        Platform = platform;
    }

    public void DelayOn(int minutes)
    {
        DelayTime += minutes;
    }

    public void DeleteDelay()
    {
        DelayTime = 0;
    }

    public override void SendNotification()
    {
        if (DelayTime > 0)
        {
            Console.WriteLine($"Установлена задержка уведомления на {DelayTime} минут.");
            Console.WriteLine("Задержка завершена.");
        }
        Console.WriteLine($"Push уведомление отправлено на платформу {Platform}.");
    }
}

EmailNotification emailNotification = new EmailNotification(
    "timalvik2017@yandex.ru", 
    1, 
    "Ваш заказ прибыл.", 
    "Входящие"
);

SMSNotification smsNotification = new SMSNotification(
    "+79504895766", 
    2, 
    "Сумма для оплаты заказа: 7777руб.", 
    "Уведомление"
);

PushNotification pushNotification = new PushNotification(
    "iOS", 
    3, 
    "Ваш заказ готов.", 
    "Уведомление"
);
pushNotification.DelayOn(30);

Notification.SendAndDisplayNotificationGroup([emailNotification, pushNotification, smsNotification]);

CorporateEmailNotification corpotateEmail = new CorporateEmailNotification("timalvik2017@yandex.com", 4, 
    "Встреча в 05.10.2025 в 12:00.", 
    "Meeting", 
    "IT"
);
corpotateEmail.SendNotification();
corpotateEmail.DisplayNotification();
```
